```python
home = '/content/drive/MyDrive/Projects/AR_CAM/App'
```

```python
import os
import shutil

# Ensure the destination directory exists (the current directory)
current_dir = os.getcwd()

# Check if the source directory exists
if os.path.exists(home):
    # List items in the source directory
    items = os.listdir(home)

    # Copy each item from the source to the current directory
    for item in items:
        src_path = os.path.join(home, item)
        dest_path = os.path.join(current_dir, item)
        try:
            if os.path.isdir(src_path):
                # continue
                # If it's a directory, copy the entire directory tree
                shutil.copytree(src_path, dest_path, dirs_exist_ok=True)
                print(f"Copied directory: {src_path} to {dest_path}")
            else:
                # If it's a file, copy the file
                shutil.copy2(src_path, dest_path)
                print(f"Copied file: {src_path} to {dest_path}")
        except Exception as e:
            print(f"Error copying {src_path}: {e}")
else:
    print(f"Source directory not found: {home}")
```

```python
import threading
import time

def run_streamlit():
    os.system('streamlit run app.py')

# Run Streamlit in a thread (Colab-safe)
thread = threading.Thread(target=run_streamlit)
thread.start()

# Wait for it to start
time.sleep(8)
```
